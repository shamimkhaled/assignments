## 📦 Step 1: Setup Directory Structure

## 📊 Step 6: View Summary Statistics

Let's see the key statistics from our analysis.

```python
if all_data:
    print("="*60)
    print("📈 SUMMARY STATISTICS")
    print("="*60)
    print(f"\n🏫 Total Campuses Processed: {len(all_data['campuses'])}")
    print(f"👥 Total Students: {all_data['total_students']}")
    print(f"⚠️  Total At-Risk Students: {all_data['total_at_risk']}")
    
    at_risk_pct = (all_data['total_at_risk'] / all_data['total_students'] * 100) if all_data['total_students'] > 0 else 0
    print(f"📊 At-Risk Percentage: {at_risk_pct:.2f}%")
    
    # Calculate overall averages
    overall_avg = calculate_subject_averages(all_data['all_students'])
    
    print("\n" + "="*60)
    print("📚 OVERALL SUBJECT AVERAGES")
    print("="*60)
    print(f"💻 Programming:       {overall_avg['programming']:.2f}")
    print(f"🗄️  Database:          {overall_avg['database']:.2f}")
    print(f"📋 Systems Analysis:  {overall_avg['systems_analysis']:.2f}")
    print(f"🌐 Networking:        {overall_avg['networking']:.2f}")
    
    # Find strongest and weakest subjects
    subjects = {
        'Programming': overall_avg['programming'],
        'Database': overall_avg['database'],
        'Systems Analysis': overall_avg['systems_analysis'],
        'Networking': overall_avg['networking']
    }
    strongest = max(subjects, key=subjects.get)
    weakest = min(subjects, key=subjects.get)
    
    print("\n" + "="*60)
    print(f"💪 Strongest Subject: {strongest} ({subjects[strongest]:.2f})")
    print(f"⚠️  Weakest Subject:  {weakest} ({subjects[weakest]:.2f})")
    print("="*60)
else:
    print("❌ No data available to display.")
```

## 🏫 Step 7: Campus-Wise Breakdown

View detailed information for each campus.

```python
if all_data:
    print("="*60)
    print("🏫 CAMPUS-WISE BREAKDOWN")
    print("="*60)
    
    for i, campus in enumerate(all_data['campuses'], 1):
        print(f"\n📍 Campus {i}: {campus['code']}")
        print("-" * 60)
        print(f"👥 Total Students: {len(campus['students'])}")
        print(f"⚠️  At-Risk Students: {len(campus['at_risk'])}")
        
        campus_risk_pct = (len(campus['at_risk']) / len(campus['students']) * 100) if len(campus['students']) > 0 else 0
        print(f"📊 At-Risk Percentage: {campus_risk_pct:.2f}%")
        
        print("\n📚 Subject Averages:")
        print(f"   💻 Programming:       {campus['averages']['programming']:.2f}")
        print(f"   🗄️  Database:          {campus['averages']['database']:.2f}")
        print(f"   📋 Systems Analysis:  {campus['averages']['systems_analysis']:.2f}")
        print(f"   🌐 Networking:        {campus['averages']['networking']:.2f}")
        
        if campus['at_risk']:
            print(f"\n⚠️  At-Risk Student IDs: {', '.join(campus['at_risk'])}")
        else:
            print("\n✅ No at-risk students in this campus!")
    
    print("\n" + "="*60)
else:
    print("❌ No data available to display.")
```

## 🔍 Step 10: Keyword Search in Report

Search for specific keywords in the academic report.

```python
def search_report(keyword):
    """Search for a keyword in the academic report."""
    report_path = 'output_files/academic_performance_report.txt'
    
    if not os.path.exists(report_path):
        print(f"❌ Report file not found at: {report_path}")
        return
    
    with open(report_path, 'r') as f:
        content = f.read()
    
    # Case-insensitive search
    count = content.lower().count(keyword.lower())
    
    print(f"🔍 Searching for: '{keyword}'")
    print(f"📊 Found {count} occurrence(s)")
    
    if count > 0:
        # Show lines containing the keyword
        lines = content.split('\n')
        matching_lines = [line for line in lines if keyword.lower() in line.lower()]
        
        print("\n📝 Matching lines:")
        print("-" * 60)
        for line in matching_lines[:10]:  # Show first 10 matches
            if line.strip():  # Only show non-empty lines
                print(line)
        
        if len(matching_lines) > 10:
            print(f"\n... and {len(matching_lines) - 10} more")
    print("\n" + "="*60)

# Example searches
print("🔎 Example Keyword Searches:\n")

search_report("Programming")
search_report("At-Risk")
search_report("Recommendations")

# You can search for your own keywords by running:
# search_report("your_keyword_here")
```

## 📥 Step 11: Download Output Files

Download all generated reports to your computer.

```python
from google.colab import files
import glob

print("="*60)
print("📥 DOWNLOADING OUTPUT FILES")
print("="*60)

# Get all output files
output_files = glob.glob('output_files/*.txt')

if output_files:
    print(f"\n📦 Found {len(output_files)} file(s) to download:\n")
    
    for file in sorted(output_files):
        filename = os.path.basename(file)
        size = os.path.getsize(file)
        print(f"   📄 {filename} ({size:,} bytes)")
    
    print("\n⏬ Starting download...\n")
    
    # Download each file
    for file in sorted(output_files):
        print(f"Downloading: {os.path.basename(file)}")
        files.download(file)
    
    print("\n✅ All files downloaded successfully!")
    print("📁 Check your browser's download folder.")
else:
    print("\n❌ No output files found.")
    print("Please run the analysis first (Steps 5-8).")

print("\n" + "="*60)
```
